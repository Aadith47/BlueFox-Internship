```python
def data_summary(df):
    rows = df.shape[0]
    data_quality = dict(df.isnull().sum())
    duplicates = df.duplicated().sum()
    about_data = df.describe()

    return dict(
        Rows=rows,
        Quality_of_data=data_quality,
        Duplicates=duplicates,
        About_data=about_data
    )


summary_of_data = data_summary(df)

print("========== DATA SUMMARY ==========")

print(f"\nRows: {summary_of_data['Rows']}")

print("\nMissing Values:")
for column, value in summary_of_data["Quality_of_data"].items():
    print(f"  {column}: {value}")

print(f"\nDuplicates: {summary_of_data['Duplicates']}")

print("\nStatistical Summary:")
print(summary_of_data["About_data"])
```

Matplotlib Basics

```python

x = np.array([1, 2, 3, 4])

y = np.array([5, 6, 2, 8])

line_custom = dict(marker = "o",
                    markersize = 10,
                    markerfacecolor = "black",
                    mec = "white",
                    linewidth = 4,
                    color = "Black")

font_custom = dict(fontsize = 20,
                   family = "Arial",
                   fontweight = "bold")

plt.xlabel("Students")

plt.ylabel("Study")

plt.title("Thanos was right", **font_custom)

plt.bar(x, y)

plt.show
```

Traveltime x Absences (Scatter)

```python

df = pd.read_csv("student-data.csv")

line_custom = dict(marker = "o",
                    markersize = 10,
                    markerfacecolor = "black",
                    mec = "white",
                    linewidth = 4,
                    color = "Black")

font_custom = dict(fontsize = 15,
                   family = "Arial",
                   fontweight = "bold")

plt.scatter(df["traveltime"], df["absences"])

plt.xlabel("Travel time", **font_custom)
plt.ylabel("Absences", **font_custom)

plt.show()
```

Pass percentage (Pie Charts)

```python
df["passed"].value_counts().plot(kind="pie", autopct="%1.2f%%")

plt.ylabel("")
plt.title("Student Pass/Fail Distribution", **font_custom)

plt.show()
```

Pass Count (Bar Charts)

```python
df["passed"].value_counts().plot(kind="bar")

plt.xlabel("Passed", **font_custom)
plt.ylabel("Number of students", **font_custom)
plt.title("Pass/Fail Count")

plt.show()
```
