```python
# import pandas as pd
# import numpy as np
# import seaborn as sns
# import matplotlib.pyplot as plt

# penguins = sns.load_dataset("penguins")

# # Head
# print("Dataset head")
# print(penguins.head())

# # Shape
# print("Shape")
# print(penguins.shape)

# # Info
# print("Info")
# penguins.info()

# # Description
# print("Description")
# print(penguins.describe())

# # Null values
# print("Null Values")
# print(penguins.isnull().sum())

# # Duplicates
# print("Duplicate Values")
# print(penguins.duplicated().sum())

# # Unique Values
# print("Unique Values")
# print(penguins.nunique())


# # Set seaborn style
# sns.set_style("whitegrid")
# sns.set_palette("husl")
# colors = ["#3D836D", "#01F08C"]

# sns.set_palette(colors)

# fig, axes = plt.subplots(3, 3, figsize=(20, 20))

# fig.suptitle(
#     "Statistical Visualization (Penguins)",
#     fontsize=25,
#     fontweight="bold"
# )


# # 1. FLIPPER LENGTH BY SPECIES AND SEX
# print("Flipper Length")

# sns.barplot(
#     data=penguins,
#     x="species",
#     y="flipper_length_mm",
#     hue="sex",
#     palette=colors,
#     ax=axes[0, 0]
# )

# axes[0, 0].set_title("Flipper Length by Species and Sex")
# axes[0, 0].set_xlabel("Species")
# axes[0, 0].set_ylabel("Flipper Length (mm)")


# # 2. FLIPPER LENGTH VS BODY MASS
# sns.scatterplot(
#     data=penguins,
#     x="flipper_length_mm",
#     y="body_mass_g",
#     palette = colors,
#     hue="species",
#     ax=axes[0, 1]
# )

# axes[0, 1].set_title("Flipper Length vs Body Mass")
# axes[0, 1].set_xlabel("Flipper Length (mm)")
# axes[0, 1].set_ylabel("Body Mass (g)")


# # 3. SPECIES COUNT BY SEX
# ax = sns.countplot(
#     data=penguins,
#     x="species",
#     hue="sex",
#     palette= colors,
#     ax=axes[0, 2]
# )

# for container in ax.containers:
#     ax.bar_label(container, fmt="%d", padding=3)

# axes[0, 2].set_title("Penguin Count by Species and Sex")
# axes[0, 2].set_xlabel("Species")
# axes[0, 2].set_ylabel("Count")


# # 4. SPECIES DISTRIBUTION
# species_count = penguins["species"].value_counts()

# explode = (0.05, 0.05, 0.05)

# axes[1, 0].pie(
#     species_count,
#     labels=species_count.index,
#     autopct="%1.1f%%",
#     explode=explode,
#     startangle=90
# )

# axes[1, 0].set_title("Penguin Species Distribution")


# # 5.BILL LENGTH VS BILL DEPTH
# sns.scatterplot(
#     data=penguins,
#     x="bill_length_mm",
#     y="bill_depth_mm",
#     hue="species",
#     palette=colors,
#     ax=axes[1, 1]
# )

# axes[1, 1].set_title("Bill Length vs Bill Depth")
# axes[1, 1].set_xlabel("Bill Length (mm)")
# axes[1, 1].set_ylabel("Bill Depth (mm)")


# # 6. CORRELATION HEATMAP
# numeric_data = penguins.select_dtypes(include="number")

# correlation = numeric_data.corr()

# sns.heatmap(
#     data=correlation,
#     annot=True,
#     cmap=colors,
#     ax=axes[1, 2]
# )

# axes[1, 2].set_title("Correlation Heatmap")


# # 7. BODY MASS BY SPECIES AND SEX
# sns.boxplot(
#     data=penguins,
#     x="species",
#     y="body_mass_g",
#     hue="sex",
#     ax=axes[2, 0]
# )

# axes[2, 0].set_title("Body Mass by Species and Sex")
# axes[2, 0].set_xlabel("Species")
# axes[2, 0].set_ylabel("Body Mass (g)")


# # 8. BILL LENGTH BY SPECIES AND SEX
# sns.boxplot(
#     data=penguins,
#     x="species",
#     y="bill_length_mm",
#     hue="sex",
#     ax=axes[2, 1]
# )

# axes[2, 1].set_title("Bill Length by Species and Sex")
# axes[2, 1].set_xlabel("Species")
# axes[2, 1].set_ylabel("Bill Length (mm)")


# # 9. BODY MASS DISTRIBUTION
# sns.histplot(
#     data=penguins,
#     x="body_mass_g",
#     kde=True,
#     bins=20,
#     ax=axes[2, 2]
# )

# axes[2, 2].set_title("Body Mass Distribution")
# axes[2, 2].set_xlabel("Body Mass (g)")
# axes[2, 2].set_ylabel("Frequency")


# plt.tight_layout()
# plt.show()




import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")

# Head
print("Dataset head")
print(penguins.head())

# Shape
print("Shape")
print(penguins.shape)

# Info
print("Info")
penguins.info()

# Description
print("Description")
print(penguins.describe())

# Null values
print("Null Values")
print(penguins.isnull().sum())

# Duplicates
print("Duplicate Values")
print(penguins.duplicated().sum())

# Unique Values
print("Unique Values")
print(penguins.nunique())


# Set seaborn style
sns.set_style("whitegrid")

# Color schemes
colors = "#3D836D"
sex_colors = {
    "Male": "#3D836D",
    "Female": "#01F08C"
}

species_colors = {
    "Adelie": "#3D836D",
    "Chinstrap": "#01F08C",
    "Gentoo": "#49A8D4"
}


fig, axes = plt.subplots(3, 3, figsize=(20, 20))

fig.suptitle(
    "Statistical Visualization (Penguins)",
    fontsize=25,
    fontweight="bold", 
    y = 1.02
)

# 1. SPECIES COUNT BY SEX
ax = sns.countplot(
    data=penguins,
    x="species",
    hue="sex",
    palette=sex_colors,
    ax=axes[0, 0]
)

for container in ax.containers:
    ax.bar_label(container, fmt="%d", padding=3)

axes[0, 0].set_title("Penguin Count by Species and Sex")
axes[0, 0].set_xlabel("Species")
axes[0, 0].set_ylabel("Count")


# 2. SPECIES DISTRIBUTION
species_count = penguins["species"].value_counts()

explode = (0.05, 0.05, 0.05)

axes[0, 1].pie(
    species_count,
    labels=species_count.index,
    autopct="%1.1f%%",
    explode=explode,
    colors=[species_colors[s] for s in species_count.index],
    startangle=90
)

axes[0, 1].set_title("Penguin Species Distribution")


# 3. BODY MASS DISTRIBUTION
sns.histplot(
    data=penguins,
    x="body_mass_g",
    kde=True,
    bins=20,
    ax=axes[0, 2]
)

axes[0, 2].set_title("Body Mass Distribution")
axes[0, 2].set_xlabel("Body Mass (g)")
axes[0, 2].set_ylabel("Frequency")


# 4. FLIPPER LENGTH BY SPECIES AND SEX
sns.barplot(
    data=penguins,
    x="species",
    y="flipper_length_mm",
    hue="sex",
    palette=sex_colors,
    ax=axes[1, 0]
)

axes[1, 0].set_title("Flipper Length by Species and Sex")
axes[1, 0].set_xlabel("Species")
axes[1, 0].set_ylabel("Flipper Length (mm)")


# 5. BODY MASS BY SPECIES AND SEX
sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="sex",
    palette=sex_colors,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Body Mass by Species and Sex")
axes[1, 1].set_xlabel("Species")
axes[1, 1].set_ylabel("Body Mass (g)")


# 6. BILL LENGTH BY SPECIES AND SEX
sns.boxplot(
    data=penguins,
    x="species",
    y="bill_length_mm",
    hue="sex",
    palette=sex_colors,
    ax=axes[1, 2]
)

axes[1, 2].set_title("Bill Length by Species and Sex")
axes[1, 2].set_xlabel("Species")
axes[1, 2].set_ylabel("Bill Length (mm)")


# 7. FLIPPER LENGTH VS BODY MASS
sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    palette=species_colors,
    ax=axes[2, 0]
)

axes[2, 0].set_title("Flipper Length vs Body Mass")
axes[2, 0].set_xlabel("Flipper Length (mm)")
axes[2, 0].set_ylabel("Body Mass (g)")


# 8. BILL LENGTH VS BILL DEPTH
sns.scatterplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    palette=species_colors,
    ax=axes[2, 1]
)

axes[2, 1].set_title("Bill Length vs Bill Depth")
axes[2, 1].set_xlabel("Bill Length (mm)")
axes[2, 1].set_ylabel("Bill Depth (mm)")


# 9. CORRELATION HEATMAP
numeric_data = penguins.select_dtypes(include="number")

correlation = numeric_data.corr()

sns.heatmap(
    data=correlation,
    annot=True,
    cmap="Greens",
    ax=axes[2, 2]
)

axes[2, 2].set_title("Correlation Heatmap")


plt.tight_layout()
plt.show()
```
