Importing and DataFrame

```python
import matplotlib.pyplot as plt

import numpy as np

import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns

df2 = pd.read_csv("student-data.csv")

df = pd.DataFrame({
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'sales': [100, 150, 120, 170, 200, 250, 180],
    'category': ['Fruit', 'Fruit', 'Veg', 'Veg', 'Dairy', 'Dairy', 'Fruit'],
    'price': [10, 15, 8, 12, 20, 18, 11]
})

line_custom = dict(marker = "o",
                    markersize = 5,
                    markerfacecolor = "black",
                    mec = "black",
                    linewidth = 3,
                    color = "blue")

font_custom = dict(fontsize = 13,
                   family = "Arial",
                   fontweight = "bold")
```

Line Charts

```python
plt.xlabel("Week",**font_custom)
plt.ylabel("Sales", **font_custom)
plt.plot(df["day"],df["sales"], **line_custom)
plt.grid(True)
plt.show()
```

Pie Charts

```python
category = df["category"].value_counts()
plt.xlabel("Category", **font_custom)
plt.pie(category, labels = category.index, autopct = "%1.1f%%")
plt.title('Pie Charts', **font_custom)
plt.show()
```

Histogram

```python

plt.hist(df["price"],bins=10, color='skyblue', edgecolor='black', density=True)
plt.xlabel('Sales')
plt.ylabel('')
plt.title('Histogram', **font_custom)
plt.show()
```

Scatter Plot

```python
plt.scatter(df["sales"], df["price"])
plt.xlabel('Sales', **font_custom)
plt.ylabel('Price', **font_custom)
plt.title('Scatter Plot', **font_custom)
plt.grid(True)
plt.show()
```

Bar Charts

```python
plt.barh(df["day"], df['sales'], color = "red")
plt.xlabel('Day', **font_custom,color = 'gray')
plt.ylabel('Sales', **font_custom, color = 'gray')
plt.title('Bar Charts', **font_custom)
```

Heatmap

```python
corr = df[['sales', 'price']].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

Box plot

```python
sns.boxplot(x='category', y='sales', data=df)
plt.xlabel("Category")
plt.ylabel("Sales")
plt.title("Sales Distribution")
plt.show()
```

Violin plot

```python
sns.violinplot(x='category', y='price', data=df)
plt.xlabel("Category")
plt.ylabel("Price")
plt.title("Price Distribution")
plt.show()
```

Pairplots

```python
sns.pairplot(df, vars=['sales', 'price'], hue='category')
plt.show()
```
